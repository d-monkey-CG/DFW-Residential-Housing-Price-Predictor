```python
import pandas as pd

df_1 = pd.read_csv("https://data.texas.gov/resource/hkzz-rrn8.csv?$limit=500000", low_memory=False)
df_2 = pd.read_csv("https://data.texas.gov/resource/cwex-zmku.csv?$limit=500000", low_memory=False)
df_3 = pd.read_csv("https://data.texas.gov/resource/vtby-uz4n.csv?$limit=500000", low_memory=False)
df_4 = pd.read_csv("https://data.texas.gov/resource/6dqt-e958.csv?$limit=500000", low_memory=False)

df = pd.concat([df_1, df_2, df_3, df_4], axis=0, ignore_index=True)
```
